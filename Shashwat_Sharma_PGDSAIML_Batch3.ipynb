# Importing Libraries

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import statsmodels.api as sm
from statsmodels.formula.api import ols
```

```python
#Uploading file
my_data = pd.read_csv(r"C:\Users\HP\Desktop\Courses Folder\Stats\Assignment\property.csv")
```

Question 1

```python
# Filtering data
altona_suburb = my_data[my_data["Suburb"] == "Altona"]
altona_prices = altona_suburb["Price"]
clean_prices = altona_prices.dropna()

# 1 sample t-test against 800000
test1 = stats.ttest_1samp(clean_prices, 800000)
q1_t_stat = test1[0]
q1_two_tailed_p = test1[1]

q1_one_tailed_p = q1_two_tailed_p / 2

print("T-Statistic:", q1_t_stat)
print("P-Value:", q1_one_tailed_p)

if q1_one_tailed_p < 0.05:
    print("Result: Null Hypothesis. The typical property price has increased.")
if q1_one_tailed_p >= 0.05:
    print("Result: Fail to reject Null Hypothesis. Price has not changed.")
```

Question 2

```python
my_data["Date"] = pd.to_datetime(my_data["Date"])

# Get only 2016 rows
data_2016 = my_data[my_data["Date"].dt.year == 2016]

# Instead of using a smart function, manually filter summer and winter months using OR conditions
# Winter is October (10), November (11), December (12), January (1), February (2), March (3)
winter_rows = data_2016[(data_2016["Date"].dt.month == 10) | 
                        (data_2016["Date"].dt.month == 11) | 
                        (data_2016["Date"].dt.month == 12) | 
                        (data_2016["Date"].dt.month == 1) | 
                        (data_2016["Date"].dt.month == 2) | 
                        (data_2016["Date"].dt.month == 3)]

# Summer is April (4) through September (9)
summer_rows = data_2016[(data_2016["Date"].dt.month == 4) | 
                        (data_2016["Date"].dt.month == 5) | 
                        (data_2016["Date"].dt.month == 6) | 
                        (data_2016["Date"].dt.month == 7) | 
                        (data_2016["Date"].dt.month == 8) | 
                        (data_2016["Date"].dt.month == 9)]

winter_prices = winter_rows["Price"].dropna()
summer_prices = summer_rows["Price"].dropna()

# Run the 2 sample test
test2 = stats.ttest_ind(summer_prices, winter_prices, equal_var=False)
print("T-Stat:", test2[0])
print("P-Value:", test2[1])

if test2[1] < 0.05:
    print("Result: Reject Null. There is a difference in prices.")
else:
    print("Result: Fail to reject Null. No difference.")
```

Question 6

```python
richmond_rows = my_data[my_data["Suburb"] == "Richmond"]
richmond_prices = richmond_rows["Price"].dropna()

q6_test = stats.ttest_1samp(richmond_prices, 1000000)

print("Q6 Null Hypothesis: Mean price in Richmond is equal to $1,000,000.")
print("Q6 Alternative Hypothesis: Mean price is not equal to $1,000,000.")
print("Q6 Test Statistic:", q6_test[0])
print("Q6 P-Value:", q6_test[1])

if q6_test[1] < 0.05:
    print("Conclusion: Reject the null hypothesis. The average price is significantly different from $1,000,000.")
else:
    print("Conclusion: Fail to reject the null hypothesis. There is no significant difference.")
```

```python
clean_data = my_data.dropna(subset=["Car", "Price"])

# Separate them out into two separate tables using basic math signs
has_car_table = clean_data[clean_data["Car"] > 0]
no_car_table = clean_data[clean_data["Car"] == 0]

prices_has_car = has_car_table["Price"]
prices_no_car = no_car_table["Price"]

test7 = stats.ttest_ind(prices_has_car, prices_no_car, equal_var=False)

print("Choice of test: Independent Two-Sample T-Test")
print("Q7 P-Value:", test7[1])
print("Interpretation: Since p-value is less than 0.05, properties with a car park sell for a higher average price.")
print("Business implication: Developers should definitely build car parking spots because people spend more money on them.")
```

Question 10

```python
clean_data = my_data.dropna(subset=["Bathroom", "Price"])

# Split into two lists based on the number 2
more_baths = clean_data[clean_data["Bathroom"] > 2]["Price"]
less_baths = clean_data[clean_data["Bathroom"] <= 2]["Price"]

test10 = stats.ttest_ind(more_baths, less_baths, equal_var=False)
q10_one_tailed_p = test10[1] / 2

print("Test choice: Independent Two-Sample T-test")
print("H0: Properties with > 2 bathrooms do not have a higher price.")
print("H1: Properties with > 2 bathrooms have a higher price.")
print("Q10 P-Value:", q10_one_tailed_p)
print("Recommendation: The claim is true. It should be noted that more bathrooms increase property values.")
```
