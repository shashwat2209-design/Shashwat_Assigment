# Question 4

```python
#-------------------Q4 i-----------
# Drop unnecessary columns if any, but let's keep relevant numerical columns
num_cols = ['Discount', 'Profit', 'Quantity', 'Sales', 'Shipping.Cost', 'Year', 'weeknum']

# Compute correlation matrix
corr_matrix = df1[num_cols].corr()
print("=== Correlation Matrix ===")
print(corr_matrix)

# Find strongest positive correlations (excluding diagonal)
corr_unstack = corr_matrix.unstack()
corr_unstack = corr_unstack[corr_unstack.index.get_level_values(0) != corr_unstack.index.get_level_values(1)]
print("\n=== Top Positive Correlations ===")
print(corr_unstack.sort_values(ascending=False).head(10))

print("Strongest Coorelation - The two features with the strongest relationship are Sales and Shipping.correlation coefficient +0.7681.")
```

```python
#-------------------Q4 ii--------------------
print("\n=== Negatively Correlated Features ===")
print(corr_unstack[corr_unstack < 0].sort_values())
print("The most prominent negative correlation in the dataset is between Discount and Profit with a coefficient of -0.3165.")
```

```python
#-------------------Q4 iii--------------------
print("=== Profit by Discount levels ===")
print(df1.groupby('Discount')['Profit'].mean())

# Let's see how profit behaves across different sales quartiles/bins
df1['Sales_Bin'] = pd.qcut(df1['Sales'], q=5, duplicates='drop')
print("\n=== Profit by Sales Quintiles ===")
print(df1.groupby('Sales_Bin', observed=False)['Profit'].mean())

print("Insight: Offering discounts is a massive driver of financial loss.When the discount is 0%, \nthe store generates an excellent average profit of $61.04 per transaction.When a discount hits 20%, \nthe average profit drops significantly to $23.55.Once discounts cross into the 30% - 50% range, \naverage profits turn negative. At extreme discount levels like 85%, the store suffers a catastrophic average loss of $1,534.33 per order.Takeaway: Price promotion strategies are severely hurting the company's bottom line.")
```

```python
#-------------------------Q4 iv-----------------
print("=== Performance Over the Years ===")
year_trends = df1.groupby('Year')[['Sales', 'Profit', 'Quantity', 'Shipping.Cost']].mean()
print(year_trends)

print("\n=== Total Volumes Over the Years ===")
year_totals = df1.groupby('Year')[['Sales', 'Profit', 'Quantity']].sum()
print(year_totals)

# Check trend over weeks to find seasonality clues
df1['week_bucket'] = pd.cut(df1['weeknum'], bins=[0, 13, 26, 39, 53], labels=['Q1', 'Q2', 'Q3', 'Q4'])
print("\n=== Quarterly/Seasonal Performance (by weeknum) ===")
print(df1.groupby('week_bucket', observed=False)[['Sales', 'Profit', 'Quantity']].mean())
print(df1.groupby('week_bucket', observed=False)[['Sales', 'Profit', 'Quantity']].sum())


print("Insight: Grouping week numbers into chronological quarters shows severe retail seasonality.")
print("Total sales volume starts at a low of $1.95 text{ Million}$ in the first weeks of the year (Q1: Weeks 1–13).")
print("Volume climbs continuously through the year, exploding to $4.41 in the final winter weeks of the year (Q4: Weeks 40–53).")
```

```python
#--------------Q5 i-----------
print("=== i. Profit by Category ===")
cat_profit_stats = df1.groupby('Category')['Profit'].agg(['median', lambda x: x.quantile(0.75) - x.quantile(0.25)]).rename(columns={'<lambda_0>': 'IQR'})
print(cat_profit_stats)

print("Highest Median Profit: Technology $29.94.\nLowest Median Profit: Office Supplies 6.55.\nWidest Spread (Largest IQR): Technology (IQR) of 98.35).\nConsistency Suggestion: The wide spread in Technology and Furniture suggests that their profits are highly inconsistent. While they contain some of the highest-paying sales, they are also prone to massive volatile swings . On the other hand, Office Supplies has very high profit consistency—it makes small, predictable, tightly-packed profits on almost every item.")
```

```python
#-------------Q5 ii------------
print("\n=== ii. Sales by Category ===")
cat_sales_stats = df1.groupby('Category')['Sales'].agg(['median'])
print(cat_sales_stats)
print("Highest Median Sales: Technology at 260.00 per transaction. \nDoes it also have the highest median profit? Yes. Technology dominates both metrics, yielding both the highest median sales and highest median profit.")
```

```python
#------------Q5 iii-------------------
print("\n=== iii. Profit by Segment ===")
seg_profit_stats = df1.groupby('Segment')['Profit'].agg(['median'])
print(seg_profit_stats)

# Most negative/low profit outliers per segment
for seg in df1['Segment'].unique():
    seg_data = df1[df1['Segment'] == seg]['Profit']
    q1 = seg_data.quantile(0.25)
    q3 = seg_data.quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    negative_outliers = seg_data[seg_data < lower_bound]
    print(f"Segment {seg}: count of low/negative profit outliers = {len(negative_outliers)}")
print("Highest Median Profit: Home Office has the highest median profit at 9.32 . \nSegment with the most negative/low profit outliers: The Consumer segment contains by far the highest number of low/negative profit outliers. This indicates that standard retail consumer orders are much more susceptible to margin-eroding discount schemes.")
```

```python
#--------------Q5 iv------------
print("\n=== iv. Sales by Segment ===")
seg_sales_stats = df1.groupby('Segment')['Sales'].agg(['median'])
print(seg_sales_stats)
print("Highest Median Sales: Consumer, Corporate, and Home Office all have an identical median sales value of 85.00.\nIs the profit pattern consistent with sales? Yes, reasonably consistent.")
```
